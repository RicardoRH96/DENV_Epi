```python
from Bio import SeqIO
from Bio.Align.Applications import MafftCommandline
import os

# 1) Paths
in_fasta  = "../Data/D2/D2gisaid_arbo_2025_04_24_18/nextclade.cds_translation.E.fasta"
out_fasta = "../Data/D2/D2gisaid_arbo_2025_04_24_18/E_Colombia.fasta"
aligned_fasta = "../Data/D2/D2gisaid_arbo_2025_04_24_18/E_Colombia_aligned.fasta"

# 2) Load & filter
filtered = []
for rec in SeqIO.parse(in_fasta, "fasta"):
    # header format assumed: ><accession>/<country>/...
    parts = rec.id.split("/")
    if len(parts) > 1 and parts[1] == "Colombia":
        filtered.append(rec)

# 3) Write filtered sequences
SeqIO.write(filtered, out_fasta, "fasta")
print(f"Wrote {len(filtered):d} records to {out_fasta!r}")

# 4) Align with MAFFT
#    (ensure 'mafft' is installed and on your PATH)
mafft_cline = MafftCommandline(input=out_fasta)
stdout, stderr = mafft_cline()

with open(aligned_fasta, "w") as handle:
    handle.write(stdout)
print(f"Alignment complete, results in {aligned_fasta!r}")
```
