```python
#load libraries
import numpy as np
import pandas as pd
import Bio
from Bio import SeqIO
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
import os
import sys
```

```python
#Read fasta and modify the header
def read_fasta(fasta_file):
    records = []
    for record in SeqIO.parse(fasta_file, "fasta"):
        record.id = record.id.split()[0]
        record.description = ""
        records.append(record)
    return records

#Write fasta
def write_fasta(records, output_file):
    with open(output_file, "w") as output_handle:
        SeqIO.write(records, output_handle, "fasta")

#Read the input file
denv2  = read_fasta('/Users/ricardorivero/Documents/DENV/2024/D2/all_colombia/subsampled_D2_filtered.fasta')
#Convert to dataframe
denv2_df = pd.DataFrame({'id': [record.id for record in denv2], 'sequence': [str(record.seq) for record in denv2]})
```

```python
# Slice the fasta header to keep the GISAID id
denv2_df['id2'] = denv2_df['id'].str.split('/').str[4]

# Convert 'id' column values to strings
denv2_df['id2'] = denv2_df['id2'].astype(str)

# Write the output file
write_fasta([SeqRecord(Seq(row['sequence']), id=row['id2'], description='') for index, row in denv2_df.iterrows()], '/Users/ricardorivero/Documents/DENV/2024/D2/all_colombia/subsampled_D2_filtered_gisaid.fasta')
```

```python
denv2_df = denv2_df.rename(columns={'id2':'gisaid_id','id':'strain'})
```

```python
denv2_df.to_csv('/Users/ricardorivero/Documents/DENV/2024/D2/all_colombia/subsampled_D2_filtered_gisaid.csv', index=False)
```
