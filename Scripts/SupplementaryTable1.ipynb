```python
import pandas as pd
import great_tables as gt
from great_tables import md
```

```python
data = pd.read_excel('../Data/SuppData/2024.09.23_DENV-2_Colombia_metadata_Genbank-Ids.xlsx')

def id_annotations(df, note, symbol: str):
    df['GenBank-ID'] = df.apply(
        lambda row: symbol if note in str(row['GenBank-Note']) else row['GenBank-ID'],
        axis=1
    )
    return df

data = id_annotations(data, note='coverage below threshold for submission', symbol='*')
data = id_annotations(data, note='co-infection', symbol='†')
data['City'] = data['City'].str.title()

data = data[['Study-ID', 'PCR-CT', 'Serotype', 'Lineage', 'Genome-Coverage', 'Collection-date', 'Country', 'City', 'GenBank-ID']]

table = gt.GT(data).tab_source_note(
    source_note='* %Coverage below the threshold for submission; † Suspected co-infection or cross-contamination'
)





```
